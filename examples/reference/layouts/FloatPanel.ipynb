The `FloatPanel` layout displays content in a floating Material UI `Paper` surface. It stays in the same viewport position while the page scrolls. Drag its empty background or padding to move it; dragging a widget inside does not move the panel.

## Parameters

### Position and content

* **`position`** (`tuple[int, int]`): `(left, top)` offset in pixels from the viewport's upper-left corner, default `(24, 24)`. Dragging updates this parameter; assigning a new tuple moves the panel.
* **`objects`** (`list`): Panel or Material UI components displayed inside the surface. Like other list-like layouts, `FloatPanel` accepts positional children and supports `append`, `extend`, and `clear`.

### Appearance

* **`elevation`** (`int`): Shadow depth of the surface when `variant="elevation"`; defaults to 1.
* **`variant`** (`"elevation"` or `"outlined"`): Use a shadow or an outline; defaults to `"elevation"`.
* **`square`** (`bool`): Remove the rounded corners; defaults to `False`.
* **`sx`** (`dict`): Material UI style overrides applied to the Paper surface.

See the [styling guide](https://panel-material-ui.holoviz.org/how_to/customize_themes_and_styles.html) for more ways to customize Material UI components.

### Floating controls

Pass widgets as positional children or through `objects`. Drag the padding around these controls; the rating and button remain interactive. The panel does not reserve space in its parent layout.

```
controls = pmui.FloatPanel(
    pmui.Rating(value=3, size="small"),
    pmui.Button(label="Submit", on_click=lambda event: print("Submitted")),
    position=(24, 80),
    elevation=4,
)

pmui.Page(title="Floating controls", main=["# Try dragging the empty surface", controls]).preview()
```

### Positioning

`position=(left, top)` is measured from the browser viewport, not from the surrounding `Page` or `Column`. Use a larger `top` offset to place the panel below a page header. The panel stays put as the page scrolls, and dragging it updates `position` in Python. Setting `position` from Python moves it again.

Focus the surface with Tab and use the arrow keys to move it in 10-pixel steps. Movement is limited to the visible viewport. In a live server app, assigning `positioned.position = (180, 130)` returns it to its starting position.

```
positioned = pmui.FloatPanel("Drag me from the padding", position=(180, 130), sx={"p": 3})

pmui.Page(title="Positioning", main=[positioned]).preview()
```

### Elevation and variants

Like `Paper`, the default `elevation` variant uses a shadow to separate the surface from the page. A higher `elevation` makes the shadow stronger. Use `variant="outlined"` for a flat surface with a border instead. `square=True` removes the rounded corners. These options follow the active Material UI theme, including dark mode.

```
shadow = pmui.FloatPanel("Elevated surface", elevation=8, position=(40, 140), sx={"p": 3})
outline = pmui.FloatPanel(
    "Outlined, square surface", variant="outlined", square=True,
    position=(300, 140), sx={"p": 3},
)

pmui.Page(title="Surface styles", main=[shadow, outline]).preview()
```

### Updating the contents

`FloatPanel` uses the same list-like API as other Material UI layouts. Add a component with `append`, replace `objects`, or call `clear` to remove them. In a live server app, calling `actions.append(...)` again updates the displayed panel without recreating it.

```
actions = pmui.FloatPanel(pmui.Button(label="First action"), position=(40, 140))
actions.append(pmui.Button(label="Second action"))

pmui.Page(title="Dynamic contents", main=[actions]).preview()
```
