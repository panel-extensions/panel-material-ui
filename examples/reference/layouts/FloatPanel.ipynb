The `pmui.FloatPanel` layout displays Panel or Material UI components in a draggable, floating Material UI `Paper` surface. The drag handle moves the panel independently of its contents, so widgets inside remain interactive.

## Parameters

### Position

* **`position`** (`tuple[int, int]`): Initial `(left, top)` position in pixels relative to the browser viewport, default `(24, 24)`. Dragging the handle updates this parameter; setting it from Python moves the panel.

### Appearance

* **`elevation`** (`int`): Shadow depth of the floating surface.
* **`square`** (`bool`): Disable rounded corners.
* **`variant`** (`"elevation"` or `"outlined"`): Surface style.
* **`sx`** (`dict`): Material UI style overrides for the surface.

The panel is positioned relative to the viewport and does not take up space in the surrounding layout. The handle can also be focused and moved with arrow keys in 10-pixel steps.

### Floating controls

Drag the handle above the controls. The rating and button can still be used without moving the panel.

```
controls = pmui.FloatPanel(
    pmui.Rating(value=3, size="small"),
    pmui.Button(label="Submit", on_click=lambda event: print("Submitted")),
    position=(24, 80),
    elevation=4,
)

pmui.Page(title="Floating controls", main=["# Try dragging the panel", controls]).preview()
```
